```python
variables = [ "taux humidite du sol", "pH du sol", "temperature", "precipitation", "humidite air", "heures ensoleillement"]
print("\nCoefficients of variation")
print(df[variables].std() / df[variables].mean())
```

```python
idx_max_precipitation = df["precipitation"].idxmax()
idx_min_precipitation = df["precipitation"].idxmin()
print("Maximum precipitation")
print(df.loc[idx_max_precipitation])
print("\nMinimum precipitation")
print(df.loc[idx_min_precipitation])
```

```python
region_stats = df.groupby("region")[[ "taux humidite du sol", "pH du sol", "temperature", "precipitation", "humidite air", "heures ensoleillement"]].agg(["mean", "median", "std", "min", "max"])
print(region_stats)
print(df["region"].value_counts())
```

```python
precipitation_summary = df.groupby("region")["precipitation"].agg(["count", "mean"])
print(precipitation_summary)
precipitation_means = precipitation_summary["mean"]
precipitation_means.plot(kind="bar",xlabel="region",ylabel="precipitation",title="Mean precipitation by region",rot=0)
plt.tight_layout()
plt.show()


```

```python
selection = df[df["precipitation"] > 100]
display(selection)
```

```python
selection2 = df[(df["precipitation"] >100 ) & (df["pH du sol"] <6 )]
display(selection2)

```

```python
plt.figure(figsize=(7, 5))
for region, group in df.groupby("region"):plt.scatter(group["precipitation"],group["pH du sol"],label=region)
plt.xlabel("precipitation")
plt.ylabel("pH du sol")
plt.title("precipitation and pH du sol")
plt.legend(title="region")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

```

```python
correlations = df[variables].corr(method="pearson")
print(correlations.round(3))
```

```python
print(df[["precipitation", "pH du sol"]].corr(method="spearman"))
```

```python
q1 = df["precipitation"].quantile(0.25)
q3 = df["precipitation"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)
print("Lower bound =", lower_bound)
print("Upper bound =", upper_bound)
outliers = df[(df["precipitation"] < lower_bound) |(df["precipitation"] > upper_bound)]
display(outliers)

```

```python
df["R"] = df["pH du sol"] / df["precipitation"]
display(df[["taux humidite du sol", "pH du sol", "temperature", "precipitation", "humidite air", "heures ensoleillement", "R"]])
idx = df["R"].idxmax()
print(df.loc[idx])

```
